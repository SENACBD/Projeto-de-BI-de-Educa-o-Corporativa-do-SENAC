# 2. Tratando e filtrando os dados

## 2.1 Normalização das colunas

```python
dados_curso = dados_curso.rename(columns={"curso": "CURSO"})
dados_curso = dados_curso.rename(columns={"Data de Inclusão do Curso": "DATA_DE_INCLUSAO_DO_CURSO"})
dados_curso = dados_curso.rename(columns={"CARGA-HORÁRIA": "CARGA_HORARIA"})
dados_curso = dados_curso.rename(columns={"SITUAÇÃO": "SITUACAO"})
dados_curso = dados_curso.rename(columns={"VALOR CONTRATADO": "VALOR_CONTRATADO"})

for column in dados_funcionario.columns:
    dados_funcionario.rename(columns={column: column.upper()}, inplace=True)

dados_participacaocurso = dados_participacaocurso.rename(columns={"MATRÍCULA": "MATRICULA"})
dados_participacaocurso = dados_participacaocurso.rename(columns={"É GESTOR?": "E_GESTOR"})
dados_participacaocurso = dados_participacaocurso.rename(columns={"LOTAÇÃO": "LOTACAO"})
dados_participacaocurso = dados_participacaocurso.rename(columns={"DATA DE INÍCIO DO CURSO": "DATA_DE_INICIO_DO_CURSO"})
dados_participacaocurso = dados_participacaocurso.rename(columns={"DATA DE CONCLUSÃO DO CURSO": "DATA_DE_CONCLUSAO_DO_CURSO"})
```

### 2.2.1 Amostra 'Curso'
---

```python
# Localizando dados nulos (esse comando retorna True para cada nulo) no dataset CURSO
dados_curso.isnull()
```

```python
# Listando a quantidade de dados nulos no dataset CURSO
dados_curso.isnull().sum()
```

```python
# Verificando valores únicos em cada coluna para encontrar valores nulos e/ou incorretos do dataset CURSO
def mostrar_valores_unicos(base, colunas=None):
    if colunas is not None:
        for coluna in colunas:
            print(f"\n{'=' * 50}")
            print(f"COLUNA: {coluna}")
            print(f"Tipo: {base[coluna].dtype}")
            print(f"Total de valores únicos: {base[coluna].nunique(dropna=False)}")
            print(f"Nulos: {base[coluna].isnull().sum()}")
            print("\nValores:")
            print(base[coluna].unique())
    else:
        for coluna in base.columns:
            print(f"\n{'=' * 50}")
            print(f"COLUNA: {coluna}")
            print(f"Tipo: {base[coluna].dtype}")
            print(f"Total de valores únicos: {base[coluna].nunique(dropna=False)}")
            print(f"Nulos: {base[coluna].isnull().sum()}")
            print("\nValores:")
            print(base[coluna].unique())


mostrar_valores_unicos(dados_curso)
```

```python
# Tratando os tipos
dados_curso["CURSO"] = dados_curso["CURSO"].astype("string")
dados_curso["EVENTO"] = dados_curso["EVENTO"].astype("string")

colunas_texto = [
    "PROGRAMA",
    "CATEGORIA",
    "FORMATO",
    "MODALIDADE",
    "SITUACAO"
]
dados_curso[colunas_texto] = dados_curso[colunas_texto].astype("string")

dados_curso["FORMATO"] = dados_curso["FORMATO"].replace("y", np.nan)

dados_curso["DATA_DE_INCLUSAO_DO_CURSO"] = pd.to_datetime(
    dados_curso["DATA_DE_INCLUSAO_DO_CURSO"],
    format="%d/%m/%Y",
    errors="coerce"
)

dados_curso["VALOR_CONTRATADO"] = (
    dados_curso["VALOR_CONTRATADO"]
    .str.replace(".", "", regex=False)
    .str.replace(",", ".", regex=False)
    .astype(float)
)

mostrar_valores_unicos(dados_curso)
```

```python
# Tratando os valores nulos
colunas = [
    "FORMATO",
    "DATA_DE_INCLUSAO_DO_CURSO",
    "PROGRAMA",
    "MODALIDADE",
    "SITUACAO"
]

# Revomendo as linhas que possuam nulos nessas colunas
dados_curso = dados_curso.dropna(subset=colunas)
```

```python
# Padronizando colunas e campos entre tabelas
dados_curso = dados_curso.rename(columns={"CURSO": "COD_CURSO"})
dados_curso = dados_curso.rename(columns={"EVENTO": "DESCRICAO_CURSO"})
dados_curso = dados_curso.rename(columns={"DATA_DE_INCLUSAO_DO_CURSO": "DATA_INCLUSAO"})
dados_curso = dados_curso.rename(columns={"SITUACAO": "CURSO_REALIZADO"})
dados_curso["CURSO_REALIZADO"] = dados_curso["CURSO_REALIZADO"].replace("Realizado", "Sim")
dados_curso["CURSO_REALIZADO"] = dados_curso["CURSO_REALIZADO"].replace("Não-Executado", "Não")
dados_curso = dados_curso.rename(columns={"VALOR_CONTRATADO": "VALOR"})

dados_curso["PROGRAMA"] = dados_curso["PROGRAMA"].replace("PDL", "Programa de Desenvolvimento de Líderes")
dados_curso["PROGRAMA"] = dados_curso["PROGRAMA"].replace("PDS", "Programa de Desenvolvimento de Sucessores")

dados_curso["CATEGORIA"] = dados_curso["CATEGORIA"].replace("TÉCNICA", "TÉCNICO")

colunas = [
    "FORMATO",
    "DATA_INCLUSAO",
    "PROGRAMA",
    "MODALIDADE",
    "CURSO_REALIZADO"
]

mostrar_valores_unicos(dados_curso, colunas)
```

## 3.4 BoxPlot para o Valores Contratados e Carga Horária

```python
def gerarBoxplots(dados_curso):

    #BoxPlot para a Coluna VALOR CONTRATADO
    fig, ax = plt.subplots()
    ax.set_ylabel('VALOR')

    for n, col in enumerate(dados_curso.columns):
        if col == 'VALOR':
            ax.boxplot(dados_curso[col], positions=[n+1])

    plt.title('Boxplot da coluna VALOR (R$)')
    plt.ylabel('Valores')
    plt.xlabel("COLUNA VALOR")
    plt.show()

    print()

    #BoxPlot para a Coluna CARGA-HORÁRIA
    fig, ax = plt.subplots()
    ax.set_ylabel('CARGA_HORARIA')

    for n, col in enumerate(dados_curso.columns):
       if col == 'CARGA_HORARIA':
            ax.boxplot(dados_curso[col], positions=[n+1])

    plt.title("BoxPlot da coluna CARGA_HORARIA = Horas ")
    plt.ylabel("CARGA-HORÁRIA")
    plt.xlabel("COLUNA CARGA_HORARIA")
    plt.show()

gerarBoxplots(dados_curso)
```

## 3.5 Carga Horária média por modalidade

```python
df_modalidade = dados_curso.groupby('MODALIDADE')[['CARGA_HORARIA']].mean().sort_values('CARGA_HORARIA')

df_modalidade.plot(kind='barh', figsize=(14, 10), color ='purple');
```
