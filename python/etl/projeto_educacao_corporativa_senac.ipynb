## 2.1 Lidando com dados nulos
---

### 2.1.1 Amostra 'Curso'
---

```python
# Localizando dados nulos (esse comando retorna True para cada nulo)
dados_curso.isnull()
```

```python
# Listando a quantidade de dados nulos
dados_curso.isnull().sum()
```

```python
# Verificando valores únicos em cada coluna para encontrar valores nulos e/ou incorretos
def mostrar_valores_unicos(base, colunas=None):
    if colunas is not None:
        for coluna in colunas:
            print(f"\n{'=' * 50}")
            print(f"COLUNA: {coluna}")
            print(f"Tipo: {base[coluna].dtype}")
            print(f"Valores únicos: {base[coluna].nunique(dropna=False)}")
            print(f"Nulos: {base[coluna].isnull().sum()}")
            print("\nValores:")
            print(base[coluna].unique())
    else:
        for coluna in base.columns:
            print(f"\n{'=' * 50}")
            print(f"COLUNA: {coluna}")
            print(f"Tipo: {base[coluna].dtype}")
            print(f"Valores únicos: {base[coluna].nunique(dropna=False)}")
            print(f"Nulos: {base[coluna].isnull().sum()}")
            print("\nValores:")
            print(base[coluna].unique())


mostrar_valores_unicos(dados_curso)
```

```python
# Tratando os tipos
dados_curso["curso"] = dados_curso["curso"].astype("string")
dados_curso["EVENTO"] = dados_curso["EVENTO"].astype("string")

colunas_texto = [
    "PROGRAMA",
    "CATEGORIA",
    "FORMATO",
    "MODALIDADE",
    "SITUAÇÃO"
]
dados_curso[colunas_texto] = dados_curso[colunas_texto].astype("string")

dados_curso["FORMATO"] = dados_curso["FORMATO"].replace("y", np.nan)

dados_curso["Data de Inclusão do Curso"] = pd.to_datetime(
    dados_curso["Data de Inclusão do Curso"],
    format="%d/%m/%Y",
    errors="coerce"
)

dados_curso["VALOR CONTRATADO"] = (
    dados_curso["VALOR CONTRATADO"]
    .str.replace(".", "", regex=False)
    .str.replace(",", ".", regex=False)
    .astype(float)
)

mostrar_valores_unicos(dados_curso)
```

```python
# Tratando os valores nulos
colunas = [
    "FORMATO",
    "Data de Inclusão do Curso",
    "PROGRAMA",
    "MODALIDADE",
    "SITUAÇÃO"
]

# Revomendo as linhas que possuam nulos nessas colunas
dados_curso = dados_curso.dropna(subset=colunas)

mostrar_valores_unicos(dados_curso, colunas)
```

### 3.4 BoxPlot para o Valores Contratados e Carga Horária

```python
def gerarBoxplots(dados_curso):

    #BoxPlot para a Coluna VALOR CONTRATADO
    fig, ax = plt.subplots()
    ax.set_ylabel('VALOR CONTRATADO')

    for n, col in enumerate(dados_curso.columns):
        if col == 'VALOR CONTRATADO':
            ax.boxplot(dados_curso[col], positions=[n+1])

    plt.title('Boxplot da coluna VALOR CONTRATADO (R$)')
    plt.ylabel('Valores')
    plt.xlabel("COLUNA VALOR CONTRATADO")
    plt.show()

    print()

    #BoxPlot para a Coluna CARGA-HORÁRIA
    fig, ax = plt.subplots()
    ax.set_ylabel('CARGA-HORÁRIA')

    for n, col in enumerate(dados_curso.columns):
       if col == 'CARGA-HORÁRIA':
            ax.boxplot(dados_curso[col], positions=[n+1])

    plt.title("BoxPlot da coluna CARGA-HORÁRIA = Horas ")
    plt.ylabel("CARGA-HORÁRIA")
    plt.xlabel("COLUNA CARGA-HORÁRIA")
    plt.show()

gerarBoxplots(dados_curso)
```

# Manipulando os dados

## Criando colunas numéricas

```python
url = 'https://raw.githubusercontent.com/millenagena/pandas-conhecendo-a-biblioteca/main/base-de-dados/aluguel.csv'
dados = pd.read_csv(url, sep=';')
dados.head()
```

```python
dados['Valor_por_mes'] = dados['Valor'] + dados['Condominio']
dados.head()
```

```python
dados['Valor_por_ano'] = dados['Valor_por_mes'] * 12 + dados['IPTU']
dados.head()
```

## Criando colunas categóricas

```python
dados['Descricao'] = dados['Tipo'] + ' em ' + dados['Bairro']
dados.head()
```

```python
dados['Descricao'] = dados['Tipo'] + ' em ' + dados['Bairro'] + ' com '  + \
                     dados['Quartos'].astype(str) + ' quarto(s) '  + \
                     ' e '  + dados['Vagas'].astype(str) + ' vaga(s) de garagem.'
dados.head()
```

```python
dados['Possui_suite'] = dados['Suites'].apply(lambda x: "Sim" if x > 0 else "Não")
dados.head()
```

```python
dados.to_csv('dados_completos_dev.csv', index=False, sep=';')
```
